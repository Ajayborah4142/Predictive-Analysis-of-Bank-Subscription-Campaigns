# *3) Data Analysis*

### ------------------------------  Customer Profile & Segmentation ---------------------------------

#### *Q.1) What is the age distribution of customers? Which age groups are more likely to subscribe (y = yes)?*

```python
data[(data["y"] == "yes")].groupby('age')['y'].value_counts().reset_index().sort_values(by='count',ascending=False).head(10)
```

#### *Q.2) Which job categories have the highest and lowest subscription rates?*

```python
job_subscription = data.groupby("job")["y"].value_counts().unstack(fill_value=0)
job_subscription

job_subscription["subscription_rate"] = (
    job_subscription["yes"] / (job_subscription["yes"] + job_subscription["no"])
) * 100

job_subscription.sort_values("subscription_rate", ascending=False)
```

#### *Q.3) Does marital status affect the likelihood of subscribing? (e.g., single vs married vs divorced).*

```python
Subcribing = data.groupby("marital")["y"].value_counts().unstack().reset_index().sort_index().sort_values(by=['no','yes'],ascending=[False,False])
Subcribing
```

#### *Q.4) How does education level influence subscription rates?*

```python
Education_subscription = data.groupby("education")["y"].value_counts().unstack(fill_value=0)
Education_subscription

Education_subscription["subscription_rate"] = (
    Education_subscription["yes"] / (Education_subscription["yes"] + Education_subscription["no"])
) * 100

Education_subscription.sort_values("subscription_rate", ascending=False)
```

### ------------------------------------ Financial Behavior --------------------------------------------

#### *Q.5) Are customers with housing loans more or less likely to subscribe?*

```python
housing_subscription = pd.crosstab(data["housing"], data["y"])
housing_subscription


housing_subscription["subscription_rate"] = (
    housing_subscription["yes"] / (housing_subscription["yes"] + housing_subscription["no"])
) * 100

housing_subscription.sort_values(by="subscription_rate",ascending=False)
```

#### *Q.6)  Does having a personal loan affect subscription probability?*

```python
data.groupby("loan")["y"].value_counts(normalize=True).unstack()

data.groupby("loan")["y"].apply(lambda x: (x=="yes").mean())
```

### *Q.7) Are customers with a credit in default more likely to avoid deposits?*

```python

data['default'] = data['default'].str.strip().str.lower()
data['y'] = data['y'].str.strip().str.lower()

# Create numerical flag for avoiding deposit
data['avoid_deposit'] = np.where(data['y'] == 'no', 1, 0)

# Group by default status
avoid_stats = (
    data.groupby('default')['avoid_deposit']
    .agg(['mean', 'count'])
    .rename(columns={'mean': 'pct_avoiding_deposit'})
)

avoid_stats['pct_avoiding_deposit'] *= 100
print(avoid_stats)
```

### ------------------------------  Business Impact & Strategy ------------------------------------------

### *Q.8)  What is the overall success rate of the campaign (percentage of y=yes)?*

```python
data['y'] = data['y'].str.strip().str.lower()

# Calculate success rate
success_rate = (data['y'].value_counts(normalize=True)['yes']) * 100
print(f"Overall success rate of the campaign: {success_rate:.2f}%")
```

### *Q.9) Which customer group (based on age, job, marital, education, and loan status) should be targeted more for higher ROI?*

```python
# Clean categorical text
for col in ['job', 'marital', 'education', 'loan', 'y']:
    data[col] = data[col].str.strip().str.lower()

# Define age groups
bins = [17, 25, 35, 45, 55, 65, 100]
labels = ['18–25', '26–35', '36–45', '46–55', '56–65', '65+']
data['age_group'] = pd.cut(data['age'], bins=bins, labels=labels)

# Group by demographic factors
target_stats = (
    data.groupby(['age_group', 'job', 'marital', 'education', 'loan'])['y']
    .value_counts(normalize=True)
    .rename('success_rate')
    .mul(100)
    .reset_index()
)

# Filter only the success (y = yes)
target_stats = target_stats[target_stats['y'] == 'yes'].sort_values('success_rate', ascending=False)

# Show top 10 high-performing groups
print(target_stats.head(10))
```

### *Q.10)	Are too many contacts reducing customer interest (over-contacting)?*

```python
data['y'] = data['y'].str.strip().str.lower()

# Group by number of contacts in current campaign
contact_effect = (
    data.groupby('campaign')['y']
    .value_counts(normalize=True)
    .rename('success_rate')
    .mul(100)
    .reset_index()
)

# Keep only y = yes (successful)
contact_effect = contact_effect[contact_effect['y'] == 'yes']
print(contact_effect.head(20))
```

# Business Insights

### 1. Customer Demographics Influence Campaign Success

* Middle-aged customers (30–50 years) showed a higher likelihood of subscribing to term deposits compared to younger or older age groups.

* Retired individuals and those with stable jobs (e.g., management, technician) had higher response rates than students or unemployed groups.

### 2. Education Level and Financial Awareness

* Customers with higher education levels (university or tertiary education) were more inclined to invest in term deposits.

* Lower-educated customers often needed more direct engagement or financial education to participate.

### 3. Previous Campaign Interaction Matters

* Customers who were previously contacted and subscribed in past campaigns were significantly more likely to respond positively again.

* High frequency of past contacts, however, reduced effectiveness—too many calls made customers less responsive.

### 4. Duration of Last Contact

* The call duration had the strongest correlation with subscription. Longer, meaningful conversations often resulted in successful conversions.

### 5. Timing and Seasonality

* Campaigns conducted in May, August, and October performed better than those in winter months.

* This could indicate higher willingness to invest during mid or post-fiscal quarters.

### 6. Economic Context

* When the employment variation rate was positive and consumer confidence index was high, customers were more willing to subscribe.

* Macroeconomic stability played a role in customer investment decisions.

### 7. Communication Channel

* Cellular contact methods (mobile phones) had a higher success rate than landlines, showing the importance of modern communication strategies.

### 8. Contact Strategy Optimization

* Reducing the number of contacts per client (avoiding overcalling) improved campaign efficiency and reduced costs.

# Business Conclusions

* Targeted marketing is more effective than broad campaigns. Focus on customers aged 30–50 with stable income and good education levels.

* Quality of interaction > quantity — meaningful conversations increase the likelihood of success.

* Campaign timing matters — launching campaigns during favorable economic periods or specific months increases conversion rates.

* Data-driven segmentation can improve ROI — using predictive models can help prioritize customers most likely to subscribe.

* Continuous improvement — analyze each campaign’s performance and refine strategies based on prior outcomes.
